# Tokenizer :
## here we will see some example of the usage of the keras library tokenizer.


````python
training_sentences = [
    "I love machine learning",
    "Machine learning is useful",
    "I love deep learning",
    "Deep learning uses neural networks",
    "Neural networks learn useful patterns",
    "Python makes machine learning practical"
]

test_sentences = [
    "I enjoy machine learning with TensorFlow",
    "Neural networks solve vision problems",
    "Python is excellent for data science"
]
````

````python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token="rafik") # <OOV> token for out-of-vocabulary words
# The older tf.keras.preprocessing.text.Tokenizer API is currently marked as deprecated;
#  TensorFlow provides TextVectorization as the modern alternative.
````

## The Modern Approach: TextVectorization
    Here is exactly how to replicate and upgrade your text preprocessing using the modern standard.
```python
import tensorflow as tf
from tensorflow.keras.layers import TextVectorization

# 1. Define the training corpus
train_data = ["hello world", "this is a test"]

# 2. Initialize the TextVectorization layer
vectorize_layer = TextVectorization(
    max_tokens=100,             # Maximum vocabulary size
    output_mode='int',          # Output integer indices
    output_sequence_length=4    # Automatically pad or truncate sentences to 4 tokens
)

# 3. Build the vocabulary (This replaces tokenizer.fit_on_texts)
vectorize_layer.adapt(train_data)

# 4. Vectorize new text containing an Out-Of-Vocabulary word
test_sentences = [
    "hello rafik",   # 'rafik' is out-of-vocabulary (not in train_data)
    "test world"
]

vectorized_output = vectorize_layer(test_sentences)

print("Vocabulary:", vectorize_layer.get_vocabulary())
print("Vectorized Sequences:\n", vectorized_output.numpy())
```
---

## Old Approach:

```python
    tokenizer.fit_on_text(training_sentences)
```
- reads all the sentences in text_data and learns the vocabulary.
- It **splits** the ``text`` into ``words``, normally converts them to ``lowercase``,
-  removes configured punctuation, counts word frequencies, 
- and assigns ``each word`` a ``numerical index``

## Here, fit_on_texts() has learned:
- which words exist,
- how often each word appears,
- and which integer should represent each word.

## It does not directly convert sentences into numbers. For that, use:
``` python
sequences = tokenizer.texts_to_sequences(text_data)

print(sequences)
```
---
## Key Differences & Handling OOV Tokens
- Automatic OOV Indexing: You no longer need to manually pass an oov_token string. 
- TextVectorization automatically reserves the string '[UNK]' (Unknown) for any out-of-vocabulary words and strictly assigns it to integer index 1.
- When the layer processes the word "rafik" in the example above, it automatically maps it to 1.

- Built-in Padding: Index 0 is completely reserved for the empty padding token (''). 
- By setting output_sequence_length, the layer pads or truncates the arrays automatically, entirely eliminating the need for the old pad_sequences() function.

- End-to-End Exporting: Because TextVectorization is an actual neural network layer, it can be passed directly as the first step in a Sequential model.
- This allows your model to accept raw text strings directly during real-world inference without needing a separate Python preprocessing script.
